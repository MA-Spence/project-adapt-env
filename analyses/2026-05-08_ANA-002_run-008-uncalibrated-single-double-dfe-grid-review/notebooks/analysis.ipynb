## Main observations

- `RUN-008` completed successfully across `81` grid settings and `324` landscapes total.
- Doubles were more deleterious than singles in every setting, and more lethal in `83.3%` of settings.
- `stability_margin` was the dominant harshness knob: lower margins sharply increased double lethality and epistasis magnitude.
- `functional_sigma_base` mostly changed the single-mutant DFE, while `epistasis_strength` mainly changed double-mutant epistasis and left single-mutant neutrality unchanged.
- The scan demonstrates broad local DFE capacity, but it is not an empirical realism test and therefore only motivates `HYP-001`.
