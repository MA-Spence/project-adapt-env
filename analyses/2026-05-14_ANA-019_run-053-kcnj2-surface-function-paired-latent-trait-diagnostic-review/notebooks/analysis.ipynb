## Main observations

- RUN-053 completed on lab-slurm as scheduler job 196 with exit code 0.
- The matched KCNJ2 surface/function readouts had Spearman 0.290 over 6789 variants.
- MAVE-NN test Spearman was 0.538 for surface trafficking and 0.389 for function, so the function observation layer was weak.
- The function-only branch reached function holdout Spearman 0.066, while the paired single-trait and explicit two-trait branches reached 0.065 and 0.021.
- The paired explicit two-trait branch had marginally lower NRMSE but worse ranking and KS than the function-only branch.
- All branches placed the reference near the peak and fitted zero epistasis, zero empirical pairwise strength, and zero noise.
- The paired state/function branches therefore did not improve the primary single-mutant function recovery package.
